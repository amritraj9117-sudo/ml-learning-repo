```python
from sklearn.metrics import make_scorer, root_mean_squared_error

def rmse_dollars(y_true_log, y_pred_log):
    return root_mean_squared_error(np.expm1(y_true_log), np.expm1(y_pred_log))

dollar_rmse_scorer = make_scorer(rmse_dollars, greater_is_better=False)
```

```python
from sklearn.preprocessing import FunctionTransformer

def add_engineered_features(X):
    X = X.copy()
    bsmt_cols = ['TotalBsmtSF', 'BsmtFullBath', 'BsmtHalfBath']
    X[bsmt_cols] = X[bsmt_cols].fillna(0)
    X['TotalSF'] = X['GrLivArea'] + X['TotalBsmtSF']
    X['TotalBaths'] = X['FullBath'] + X['BsmtFullBath'] + 0.5 * (X['HalfBath'] + X['BsmtHalfBath'])
    X['TotalPorchSF'] = X['WoodDeckSF'] + X['OpenPorchSF'] + X['EnclosedPorch'] + X['3SsnPorch'] + X['ScreenPorch']
    X['HouseAge'] = X['YrSold'] - X['YearBuilt']
    X['RemodelAge'] = X['YrSold'] - X['YearRemodAdd']
    X['HasPoll'] = (X['PoolArea'] > 0).astype(int)
    X['QualArea'] = X['OverallQual'] * X['GrLivArea']
    return X

def add_engineered_features_out(transformer, input_features):
    new_cols = ['TotalSF', 'TotalBaths', 'TotalPorchSF', 'HouseAge', 'RemodelAge', 'HasPoll', 'QualArea']
    return np.array(list(input_features) + new_cols)

feature_engineer = FunctionTransformer(
    func=add_engineered_features,
    feature_names_out=add_engineered_features_out
)
```

```python
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer


cat_none_cols = ['Alley', 'BsmtQual', 'MasVnrType', 'BsmtExposure', 'BsmtCond', 
                 'BsmtFinType1', 'BsmtFinType2', 'PoolQC', 'Fence', 'MiscFeature', 'GarageQual', 
                 'GarageCond', 'GarageType', 'GarageFinish', 'FireplaceQu']
cat_mode_cols = [
    'Electrical',
    'Street', 'LotShape', 'Utilities', 'LandSlope', 'HeatingQC', 'ExterQual', 'KitchenQual', 'ExterCond',
    'MSZoning', 'MSSubClass', 'LandContour', 'LotConfig', 'Condition2', 'BldgType', 'HouseStyle',
    'RoofStyle', 'RoofMatl', 'Exterior1st', 'Exterior2nd', 'Heating', 'CentralAir', 'Functional',
    'PavedDrive', 'SaleType', 'SaleCondition', 'Foundation'
]
num_median_cols = ['LotFrontage']
num_zero_cols = ['MasVnrArea', 'GarageYrBlt', 'GarageCars', 'GarageArea', 'BsmtFinSF1', 'BsmtFinSF2', 'BsmtUnfSF']

imputer_none = SimpleImputer(strategy='constant', fill_value='None',keep_empty_features=True)
imputer_mode = SimpleImputer(strategy='most_frequent')
imputer_median = SimpleImputer(strategy='median')
imputer_zero = SimpleImputer(strategy='constant', fill_value=0.0)

filling_preprocessor = ColumnTransformer(
    transformers=[
        ('cat_none', imputer_none, cat_none_cols),
        ('cat_mode', imputer_mode, cat_mode_cols),
        ('num_median', imputer_median, num_median_cols),
        ('num_zero', imputer_zero, num_zero_cols)
    ],
    remainder='passthrough',
    verbose_feature_names_out=False 
)
```

```python
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import TargetEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

street_order = ['Grvl', 'Pave']
alley_order = ['None', 'Grvl', 'Pave']
lotshape_order = ['IR3', 'IR2', 'IR1', 'Reg']
utilities_order = ['ELO', 'NoSeWa', 'NoSewr', 'AllPub']
landslope_order = ['Gtl', 'Mod', 'Sev']
BsmtQual_order = ['Ex', 'Gd', 'TA', 'Fa', 'Po', 'None']
BsmtCond_order = ['Ex', 'Gd', 'TA', 'Fa', 'Po', 'None']
BsmtExposure_order = ['Gd', 'Av', 'Mn', 'No', 'None']
BsmtFinType1_order = ['GLQ', 'ALQ', 'BLQ', 'Rec', 'LwQ', 'Unf', 'None']
BsmtFinType2_order = ['GLQ', 'ALQ', 'BLQ', 'Rec', 'LwQ', 'Unf', 'None']
HeatingQC_order = ['Ex', 'Gd', 'TA', 'Fa', 'Po']
ExterQual_order = ['Ex', 'Gd', 'TA', 'Fa', 'Po']
KitchenQual_order = ['Ex', 'Gd', 'TA', 'Fa', 'Po']  
FireplaceQu_order = ['Ex', 'Gd', 'TA', 'Fa', 'Po', 'None']
GarageQual_order = ['Ex', 'Gd', 'TA', 'Fa', 'Po', 'None']
GarageCond_order = ['Ex', 'Gd', 'TA', 'Fa', 'Po', 'None']
PoolQC_order = ['Ex', 'Gd', 'TA', 'Fa', 'None']
ExterCond_order = ['Ex', 'Gd', 'TA', 'Fa', 'Po']

ordinal_cols = ['Street', 'Alley', 'LotShape', 'Utilities', 'LandSlope', 'BsmtQual', 'BsmtCond', 'BsmtExposure',
                 'BsmtFinType1','BsmtFinType2', "HeatingQC", "ExterQual","KitchenQual","FireplaceQu",
                   "GarageQual", "GarageCond", "PoolQC","ExterCond"] 
categorical_cols = ['MSZoning', 'MSSubClass', 'LandContour', "LotConfig", "Condition2","BldgType", "HouseStyle","RoofStyle","RoofMatl", "Exterior1st",
                     "Exterior2nd","MasVnrType","Heating", "CentralAir", "Electrical", "Functional", "GarageType", "GarageFinish", "PavedDrive", 
                     "Fence","MiscFeature", "SaleType", "SaleCondition", "Foundation" ]
target_columns = ['Neighborhood', 'Condition1']

encoding_preprocessor = ColumnTransformer(
    transformers=[
        ('ordinal', OrdinalEncoder(categories=[street_order, alley_order, lotshape_order, utilities_order, landslope_order,BsmtQual_order,BsmtCond_order,BsmtExposure_order,
                                               BsmtFinType1_order,BsmtFinType2_order,HeatingQC_order,ExterQual_order,KitchenQual_order,FireplaceQu_order,GarageQual_order,
                                               GarageCond_order, PoolQC_order, ExterCond_order]), ordinal_cols),
        ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False), categorical_cols),
        ('target', TargetEncoder(target_type='continuous'), target_columns)
    ],
    remainder='passthrough',
    verbose_feature_names_out=False
)

```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import FunctionTransformer
log_cols = ['GrLivArea', 'TotalSF', '1stFlrSF', 'TotalBsmtSF', 'LotArea', 'MasVnrArea']

standard_scale_cols = ['OverallQual', 'GarageCars', 'TotalBaths', 'GarageArea', 'FullBath', 
                       'TotRmsAbvGrd', 'YearBuilt', 'YearRemodAdd', 'GarageYrBlt', 'Fireplaces', 
                       'LotFrontage', 'WoodDeckSF', 'OpenPorchSF']


log_and_scale_pipe = Pipeline(steps=[
    ('log1p', FunctionTransformer(np.log1p, feature_names_out='one-to-one')),
    ('scaler', StandardScaler())
])
scaling_preprocessor = ColumnTransformer(
    transformers=[
        ('log_and_scale', log_and_scale_pipe, log_cols),
        ('std_scale', StandardScaler(), standard_scale_cols)
    ],
    remainder=StandardScaler(),
    verbose_feature_names_out=False
)
```

```python

prepeocessing = Pipeline(steps=[
    ('feature_engineering', feature_engineer),
    ('imputation', filling_preprocessor),
    ('encoding', encoding_preprocessor),
    ('scaling', scaling_preprocessor)
])
prepeocessing.set_output(transform="pandas")
```

```python
X_transformed = prepeocessing.fit_transform(train_X, train_y)
```
